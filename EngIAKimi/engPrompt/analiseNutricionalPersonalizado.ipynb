```python
from dotenv import load_dotenv
from groq import Groq
import os
load_dotenv()

client = Groq(
    api_key=os.environ.get("GROQ_API"),  # This is the default and can be omitted
)
```

```python
system_prompt = """
Voce é um especialista em nutrição e receitas FIT e precisa avaliar de forma resumida a solicitação de receitas para melhor
ajustar no paladar das pessoas com as seguintes caracteristicas:
1. gosta de doce e precisa diminuir gradualmente
2. gosta muito de leite em po e se estiver afetando a perde de peso saber dimuir gradualmente
3. gosta muito de queijo e se estiver afetando a perde de peso saber dimuir gradualmente
4. precisa ajustar da melhor forma sua alimentacao de acordo com os pontos que forem passados
5. precisa perder peso por dia para melhor visualizacao do resultado
6. precisa saber o tempo medio por dia para alimentacao e exercicio para perder pelo menos 1kilo por dia, ou o maximo possivel que nao acabe com a saude
7. precisa de formas de mensagens e alimentos de facil aquisicao e custos acessiveis
8. receitas para todos os dias da semana mas q sab e dom seja opcional
9. nao separar os ovos das gemas acho despedicio, receitas devem conter ovo inteiro
10. buscar manter os ingredientes e so ir alterando as poucos deixar separado a a alteracoes gradativa.
11. ajustar o q ja é adquirido com opcoes ja usadas com opcoes diferente entre parenteses
12. primeiro trazer uma avaliacao do que ja tem e dps o que pode melhorar gradualmente
13. em cada ponto de avaliacao precisa colocar um descricao como sugestao de melhoria e justificativa se caso seja uma avaliacao negativa
14. as avaliacao devem ser feitas em topicos por ingredientes e se por periodo fazer subtopicos para menter uma visualizacao melhor do texto
    (
        Exemplo:
        Cafe da manha:
            1. Cafe expresso - ajusta para tal coisa, pq isso e aquilo
            3. Leite em po - precisa diminui pq tal e o recomendado é tal
    )

Obs: Formate o resultado em uma tabela Markdown. Certifique-se de que o código visual da tabela esteja perfeitamente alinhado e indentado, utilizando espaços em branco para que as barras verticais (|) fiquem retas, facilitando a leitura do código bruto.
"""

user_prompt = """De acordo com a lista de alimentos adquiridos diariamente fazer novas receitas de acordo com os requisitos:
              1. cafe da manha se for comer antes das 10 seria uma banana e quebraria o jejum mas apenas qdo vou caminhar ou fazer exercicio, e apos as 10 seria cafe especial expresso com quase 5 a 10 g de leite e 5 a 8 g de acucar demerara com um wrap10 com queijo e frango ou queijo e presunto
              2. no almoco arroz, macarrao salada e frango ou carne chega em torno de 800g tudo... o arroz tem precurado diminuir cada vz mais
              3. estou aplicando o jejum das 20 as 10(manter se necessario ou ajustar diminuindo o periodo ou ajustando os horarios se necessario tb)
              4. a tarde as vzs so um cafe(expresso com acucar e leite q nem no cafe da manha) e as vzs com torrada ou biscoito cream cracker com queijo tb
              5. antes de dormir(ate as 20) as vzs um wrap 10 q nem no cafe da manha ou crepioca, ou ovos(5 ovos cozidos), ou frango cozido, ou pede um japones q geralmente com cream cheese
              6. as vzs antes das 20 um minguau feito com aveia/neston/farinha lactea.. ou os 3 juntos... com leite em po... as vz mais leite em po q eles p ficar bem consistente... umas 400g total + ou -
              """

assist_prompt =  """| Situacao      | alimento | valido |\n
                    | :-----------  | :----    | :---   |\n
                    | Almoco        | arroz    | sim    |\n
                    |               | frango   | nao    |\n
                    | Jantar        | ovo      | talvez |"""
response = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
        #{"role": "assistant", "content" : assist_prompt}
    ],
    temperature=0.5
)
print("Resposta:", response.choices[0].message.content)
```
